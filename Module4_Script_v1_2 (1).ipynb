## Megatelco Data Dictionary

 DEMOGRAPHIC VARIABLES:
 - College - has the customer attended some college (one, zero)
 - Income - annual income of customer
 - House - estimated price of the customer's home (if applicable)

 USAGE VARIABLES:
 - Data Overage Mb - Average number of megabytes that the customer used in excess of the plan limit (over last 12 months)
 - Data Leftover Mb - Average number of megabytes that the customer use was below the plan limit (over last 12 months)
 - Data Mb Used - Average number of megabytes used per month (over last 12 months)
 - Text Message Count - Average number of texts per month (over last 12 months)
 - Over 15 Minute Calls Per Month - Average number of calls over 15 minutes in duration per month (over last 12 months)
 - Average Call Duration- Average call duration (over last 12 months)

PHONE VARIABLES:
 - Operating System - Current operating system of phone
 - Handset Price - Retail price of the phone used by the customer

ATTITUDINAL VARIABLES:
 - Reported Satisfaction - Survey response to "How satisfied are you with your current phone plan?" (high, med, low)
 - Reported Usage Level - Survey response to "How much do your use your phone?" (high, med, low)
 - Considering Change of Plan - Survey response to "Are you currently planning to change companies when your contract expires?" (high, med, low)

OTHER VARIABLES
 - Leave - Did this customer churn with the last contract expiration? (LEAVE, STAY)
 - ID - Customer identifier

## Clean up the data

## Calculate distribution of the outcome  
What is the proportion of people who churned?

Why should we care about this proportion?

1. An important step in EDA is to understand the distribution of the target variable.

2. The majority class in the target variable will serve as an important benchmark for model performance. If we used what we'll call a "majority class classifier"---this consists in always predicting the majority class, which in this case is `STAY`---we would be correct 1 - .49 or 51% of the time.  Another way of saying this is that majority class classifier in the MegaTelCo case would result in accuracy of .51.  

Accuracy is defined as the proportion of correctly predicted labels. It is a commonly used error metric for evaluating classifier performance.

Think about why a majority class model in this case would have an accuracy of .51.

Whatever later model we develop should have better accuracy than this performance benchmark.

```python
#Add new field with binary value for leave
df_clean['leave_flag'] = df_clean['leave'].str.replace('STAY','0')
df_clean['leave_flag'] = df_clean['leave_flag'].str.replace('LEAVE','1')

#Convert to integer
df_clean['leave_flag'] = df_clean['leave_flag'].astype('int')

#Find the mean value
df_clean['leave_flag'].mean()
```

```python

# Explanation of why this proportion is important
print("\nThis proportion is crucial for several reasons:")
print("1. Establishing a baseline for model performance:  The majority class's proportion acts as a benchmark for any predictive model's performance.  In this case, since the majority did not churn (STAY), a simple 'majority class classifier' that always predicts 'STAY' would be correct in a percentage equal to (1 - churn proportion). This accuracy value is the baseline against which more sophisticated models must compete.")
print("2. Understanding the class imbalance: The proportion reveals whether we have a balanced dataset or an imbalanced one. If one class is significantly more prevalent than others (as is the case here with 51% not churning and 49% churning), it can lead to biased model evaluation, which is why it is important to understand the distribution of our target variable.")
print("3. Informing model selection and evaluation: Highly imbalanced datasets require specific strategies during model development and evaluation. Metrics like accuracy might not be appropriate, and other metrics such as precision, recall, F1-score, and AUC-ROC might become more useful.")
print("4. Assessing business impact: The churn proportion helps quantify the scale of the problem, allowing businesses to determine the resources to invest in improving customer retention.")
```

```python
# Bar chart for Income_Level vs. Churn
plt.figure(figsize=(10, 6))
df_clean.groupby('Income_Level', observed=True)['leave_flag'].mean().plot(kind='bar')
plt.title('Churn Rate by Income Level')
plt.xlabel('Income Level')
plt.ylabel('Churn Rate')
plt.show()
```

```python
# Histogram for Handset Price vs. Churn
plt.figure(figsize=(10, 6))
plt.hist(df_clean[df_clean['leave_flag'] == 0]['handset_price'], alpha=0.5, label='Stayed')
plt.hist(df_clean[df_clean['leave_flag'] == 1]['handset_price'], alpha=0.5, label='Left')
plt.title('Handset Price Distribution by Churn')
plt.xlabel('Handset Price')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

```python
# Scatter plot for Data MB Used vs. Text Message Count, colored by Churn
plt.figure(figsize=(10, 6))
plt.scatter(df_clean[df_clean['leave_flag'] == 0]['data_mb_used'],
            df_clean[df_clean['leave_flag'] == 0]['text_message_count'],
            label='Stayed', alpha=0.5)
plt.scatter(df_clean[df_clean['leave_flag'] == 1]['data_mb_used'],
            df_clean[df_clean['leave_flag'] == 1]['text_message_count'],
            label='Left', alpha=0.5)
plt.title('Data Usage vs. Text Messages by Churn')
plt.xlabel('Data MB Used')
plt.ylabel('Text Message Count')
plt.legend()
plt.show()
```
